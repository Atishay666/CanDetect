```python
df["Class"].value_counts()      ##0 means genuine transaction & 1 means fraud transaction
```

```python
import seaborn as sns
```

```python
##Fraud vs Genuine Transations
plt.figure(figsize=(6,4))
sns.countplot(x='Class', data=df)

plt.title("Distribution of Fraud and Genuine Transactions")
plt.xlabel("Class")
plt.ylabel("Number of Transactions")

plt.show()
```

```python
##Percentage of Fraud Transactions
fraud_percentage = (df["Class"].value_counts(normalize=True) * 100)

print(fraud_percentage)
```

```python
plt.figure(figsize=(6,6))

plt.pie(df["Class"].value_counts(),
        labels=["Genuine","Fraud"],
        autopct="%1.2f%%")

plt.title("Percentage of Fraud Transactions")

plt.show()
```

```python
df["Amount"].describe()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df["Amount"], bins=50)

plt.title("Distribution of Transaction Amount")

plt.xlabel("Amount")

plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Class", y="Amount", data=df)

plt.title("Transaction Amount by Class")

plt.show()
```

```python
plt.figure(figsize=(15,12))

correlation = df.corr()

sns.heatmap(correlation, cmap="coolwarm")

plt.title("Correlation Matrix")

plt.show()
```
